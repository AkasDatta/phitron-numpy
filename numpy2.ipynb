# Array creation with Random Values  

```python
# np.random.uniform(start, end, shape)

arr = np.random.uniform(50, 100, (3,4))

print(arr)

print(type(arr))
print(arr.dtype)

print(arr.shape)
print(arr.ndim)
print(arr.size) 
```

```python
# np.random.rand(shape)
import numpy as np
arr = np.random.rand(2,3)

print(arr)

print(type(arr))
print(arr.dtype)

print(arr.shape)
print(arr.ndim)
print(arr.size) 
```

```python
# np.random.randint(shape)

arr = np.random.randint(2,3, (2,3))

print(arr)

print(type(arr))
print(arr.dtype)

print(arr.shape)
print(arr.ndim)
print(arr.size) 
```

```python
# np.random.randint(shape)

arr = np.random.randint(1,10, (10,10))

print(arr)

print(type(arr))
print(arr.dtype)

print(arr.shape)
print(arr.ndim)
print(arr.size) 
```

```python
# np.random.uniform(start, end, shape)

arr = np.random.uniform(50, 100, (3, 3, 4))

print(arr)

print(type(arr))
print(arr.dtype)

print(arr.shape)
print(arr.ndim)
print(arr.size) 
```
