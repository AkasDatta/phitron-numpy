# array creation with range 

# np.arange(first, end, step ) -> similar top range function 

arr = np.arange(1, 10, 2)

print(arr)

print(type(arr))
print(arr.dtype)

print(arr.shape)
print(arr.ndim)
print(arr.size) 

```python


arr = np.arange(1, 10, 1).reshape(3,3)
mat = arr.reshape(3,3)

print(arr)
print(mat)
print(type(arr))
print(arr.dtype)

print(arr.shape)
print(arr.ndim)
print(arr.size) 
```

```python


arr = np.linspace(1, 10, 11)

print(arr)

print(type(arr))
print(arr.dtype)

print(arr.shape)
print(arr.ndim)
print(arr.size) 
```

```python
# np.arange(start, end(included), koita )

arr = np.logspace(1, 4, 5)
arr2 = np.logspace(0, 4, 5)

print(arr)
print(type(arr))
print(arr.dtype)

print(arr.shape)
print(arr.ndim)
print(arr.size) 
print(arr2)
print(type(arr2))
print(arr2.dtype)

print(arr2.shape)
print(arr2.ndim)
print(arr2.size) 
```
