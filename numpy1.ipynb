```python
import numpy as np

# 1d array 

arr1 = np.array([1,2,3,4,5])
print(arr1.ndim)

# 2d array 
arr2 = np.array([ [1, 2, 3], #row
                  [4, 5, 6]  #row2
                ])  

print(arr2.ndim)

# 3d array
arr3 = np.array([
    # 1st floor 
      [ [1, 2, 3], #row
        [4, 5, 6]  #row2
      ],
    

       # 2nd floor 
      [ [1, 2, 3], #row
        [4, 5, 6]  #row2
      ]
    
    ])  

print(arr3.ndim)

# again floor 
arr3 = np.array([
    # 1st floor 
      [ [1, 2, 3], #row
        [4, 5, 6]  #row2
      ],
    

       # 2nd floor 
      [ [1, 2, 3], #row
        [4, 5, 6]  #row2
      ],
    

       # 3rd floor 
      [ [1, 2, 3], #row
        [4, 5, 6]  #row2
      ]
    
    ])  

print(arr3.ndim)
```

```python
# tuple 

tpl = (10, 20, 30)
arr = np.array(tpl)
arr = np.array(tpl,dtype=np.int32)
print(arr.dtype)
arr = arr.astype(np.int16)


print(arr)
print(type(arr))
print(arr.dtype)

print(arr.shape)
print(arr.ndim)
```

```python
# set

st = (10, 20, 30)
arr = np.array(st)
arr = np.array(st,dtype=np.int32)
print(arr.dtype)
arr = arr.astype(np.int16)


print(arr)
print(type(arr))
print(arr.dtype)

print(arr.shape)
print(arr.ndim)


# set 2nd 

st = (10, 20, 30)
arr2 = np.array(list(st))
arr2 = np.array(st,dtype=np.int32)
print(arr2.dtype)
arr2 = arr2.astype(np.int16)


print(arr2)
print(type(arr2))
print(arr2.dtype)

print(arr2.shape)
print(arr2.ndim)
```
