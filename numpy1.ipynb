# numpy = Numarical python 

```python
arr = [1, 2, 3]
print(arr * 5)

arr = [x * 5 for x in arr]
print(arr)
```

```python
import numpy as np

arr1 = np.array([1, 2, 3])
arr1 = arr1 * 5

print(arr1)
```

## NDarray

```python
import numpy as np

# 1d array 

arr1 = np.array([1,2,3,4,5])

print(type(arr1.ndim))

# 2d array 
arr2 = np.array([1, 2, 3,], #row
                [4, 5, 6])  #row2

print(arr2.ndim)
```

```python
import numpy as np

# 1d array 

arr1 = np.array([1,2,3,4,5])
print(arr1.ndim)

# 2d array 
arr2 = np.array([ [1, 2, 3], #row
                  [4, 5, 6]  #row2
                ])  

print(arr2.ndim)

# 3d array
arr3 = np.array([
    # 1st floor 
      [ [1, 2, 3], #row
        [4, 5, 6]  #row2
      ],
    

       # 2nd floor 
      [ [1, 2, 3], #row
        [4, 5, 6]  #row2
      ]
    
    ])  

print(arr3.ndim)

# again floor 
arr3 = np.array([
    # 1st floor 
      [ [1, 2, 3], #row
        [4, 5, 6]  #row2
      ],
    

       # 2nd floor 
      [ [1, 2, 3], #row
        [4, 5, 6]  #row2
      ],
    

       # 3rd floor 
      [ [1, 2, 3], #row
        [4, 5, 6]  #row2
      ]
    
    ])  

print(arr3.ndim)
```
