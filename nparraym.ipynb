```python
# comparison - >, < >=, <=, !=, == 

x= np. array([10, 8, 2, 67])
y= np. array([30, 2, 87, 100])

greater_than = x<y
print(greater_than)

equal = x==y
print(equal)

# all() -> shob true kina
# any() - jekono ekta true ase kina 

print(np.all(greater_than))
print(np.any(equal))
```

```python
# searching 
print(x)

# np.where(condition, x, y) if true x, otherwise replace with y
index = np.where(x == 16) #returen index 
print(index)

arr = np.where(x > 16) #returns array 
print(arr)
arr = np.where(x > 16, x,0)
print(arr)
```
